Quantity: The quantities of each product (item) per transaction. Numeric.

CustomerID: Customer number. Nominal. A 5-digit integral number uniquely assigned to each customer.

#### Filtering Data###

```python
clean_data = data.copy()
```

```python
clean_data = clean_data.drop_duplicates()

clean_data = clean_data[clean_data['Quantity'] > 0]

clean_data = clean_data[clean_data['Price'] > 0]

clean_data = clean_data.dropna(subset=['Customer ID'])

clean_data['Revenue'] = (
    clean_data['Quantity'] * clean_data['Price']
)
```

## 6. Exploratory Data Analysis (EDA)

### 6.1 Monthly Revenue Analysis

We analyze monthly revenue to identify:
- Sales trends
- Seasonal patterns
- High/low performing periods

```python
monthly_sales = (
    clean_data
    .groupby(clean_data['InvoiceDate'].dt.to_period('M'))['Revenue']
    .sum()
)

monthly_sales
```

```python
monthly_sales.sort_values(ascending=False).head(5)
```

```python
monthly_sales.sort_values().head(5)
```

```python
country_sales = (
    clean_data
    .groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

country_sales.head(10)
```

```python
product_sales = (
    clean_data
    .groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

product_sales.head(10)
```

```python
customer_sales = (
    clean_data
    .groupby('Customer ID')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

customer_sales.head(10)
```
